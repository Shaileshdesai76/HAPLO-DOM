```python
class TreeNode:
    def __init__(self, name: str, defining_mutations=None):
        self.name = name
        self.defining_mutations = (defining_mutations)   # Use set for fast lookup
        self.children = []                                        # Flexible list of children
        self.parent = None                                        # Optional: useful for backtracking

    def add_child(self, child_node):
        child_node.parent = self
        self.children.append(child_node)
        return child_node
```

```python
import ast

def read_mutation_file(filename="mutation_results.txt"):

    mutation_dict = {}
    result_dict = {}

    threshold = 0.06

    print("--- Reading and Parsing File ---")

    # Read mutation file
    with open(filename, "r") as file:

        for line in file:

            line = line.strip()

            if not line or ":" not in line:
                continue

            sample_id, list_string = line.split(":", 1)

            sample_id = sample_id.strip()
            list_string = list_string.strip()

            mutation_list = ast.literal_eval(list_string)

            # Store mutations as a set
            mutation_dict[sample_id] = set(mutation_list)

    print(f"Loaded {len(mutation_dict)} samples")

    # Process each sample independently
    for sample_id, sample_mutations in mutation_dict.items():

        current = root     # IMPORTANT: start from root for each sample

        while current.children:

            best_child = None
            best_score = 0

            # Evaluate all children
            for child in current.children:

                if len(child.defining_mutations) == 0:
                    continue

                child_mutations = set(child.defining_mutations)

                matches = sample_mutations & child_mutations

                score = len(matches) / len(child_mutations)

                # Optional debugging
                print(
                    f"{sample_id} | "
                    f"{current.name} -> {child.name} | "
                    f"{len(matches)}/{len(child_mutations)} = {score:.3f}"
                )

                if score > best_score:
                    best_score = score
                    best_child = child

            # No suitable child found
            if best_child is None:
                break

            # Move down tree if threshold passed
            if best_score >= threshold:
                current = best_child
            else:
                break

        # Store final assigned haplogroup
        result_dict[sample_id] = current

        with open ("result.out", "w+") as file:
          for sample, haplo in result_dict.items():
            file.write(f"{sample} -> {haplo.name}\n")

    return result_dict

results = read_mutation_file()

for sample, haplo in results.items():

    print(sample, "->", haplo.name)
```
